```python
class BankAccount:
    def __init__(self,balance=0):
        self.__balance = balance
    def deposit(self,amt1):
        self.__balance += amt1
        return self.__balance
    def withdraw(self,amt2):
        self.__balance -= amt2
        return self.__balance
    def display_balance(self):
        return self.__balance
acc1 = BankAccount(2000)
print(acc1.deposit(200))
print(acc1.withdraw(50))
print(acc1.deposit(200))
print(acc1.deposit(100))   
print(acc1.withdraw(150))
print(acc1.display_balance())    
```

```python
class Employee:
    def __init__(self,name,salary):
        self.salary = salary
        self.name = name
    def salarygot(self):
        return self.salary
        
class Manager(Employee):
    def __init__(self,name,salary,bonus):
        super().__init__(name,salary)
        self.bonus = bonus
    def addbonus(self):
        self.salary += self.bonus
        return self.salary    
   
s1 = Manager("sidhharth",40000,2000)
s1.addbonus()
print(s1.salarygot())
```

```python
class Vehicle:
    def start(self):
        print("Started...")
        return "Vehicle started"
class Car(Vehicle):
    def show_model(self, model):
        super().start()
        self.model = model
        return f"Model: {self.model}"
v1 = Car()
print(v1.show_model("Rangerover"))
```

```python
class Teacher:
    def __init__(self, marks):
        self.marks = marks
    def show_Teacher_marks(self):
        print(f"Teacher Marks: {self.marks}")
class Sports:
    def __init__(self, score):
        self.score = score
    def show_sports_score(self):
        print(f"Sports Score: {self.score}")
class Result(Teacher, Sports):
    def __init__(self, marks, score):
        Teacher.__init__(self, marks)
        Sports.__init__(self, score)
    def total_result(self):    
        total_perform = self.marks + self.score
        return total_perform
r1 = Result(81, 84)        
r1.show_Teacher_marks()  
r1.show_sports_score()  
print(f"Total Result: {r1.total_result()}")
```
