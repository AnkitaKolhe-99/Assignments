```python
def productofnum(num):
    p = 1
    while num != 0:
        rem = num % 10
        p = p * rem
        num = num // 10
    return p
result = productofnum(45)
print(result)
```

```python
def factors_while(num):    
    i = 1
    while num != 0:
        rem = num % i
        i = i + 1
    return rem    
result = factors_while(9)
print(result)
```
