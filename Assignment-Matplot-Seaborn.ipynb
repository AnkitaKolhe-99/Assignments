```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [10,20,15,25,30]
a = np.array(x)
b = np.array(y)
plt.plot(a,b,color='red',linestyle='dashed')
plt.title("Line graph")
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

```python
a = ['Jan','Feb','Mar','Apr']
b = [200,250,300,280]
x = np.array(a)
y = np.array(b)
plt.bar(x,y,color='green')
plt.show()
```

```python
x = np.random.randint(1,200,50)
y = np.random.randint(1,100,50)
plt.scatter(x,y,color='red')
plt.show()
```

```python
x = np.random.normal(10,99,100)
plt.hist(x,color='orange')
plt.show()
```

```python
y1 = [1,4,9,16,25]
y2 = [2,5,10,17,26]
x = np.array(y1)
y = np.array(y2)
plt.plot(x,linestyle='dashed',color='blue')
plt.plot(y,linestyle='dotted',color='green')
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("2 Lines graph")
plt.show()
```

```python
x = [40,30,20,10]
l = ['A','B','C','D']
c = ['green','blue','red','yellow','orange']
e = [0.03,0,0,0]
a = np.array(x)
plt.pie(a, labels=l, colors=c, explode=e)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x_values = ['A', 'B', 'C', 'D', 'E']
y_line = [10, 15, 7, 12, 14]
y_bar = [20, 35, 30, 45, 40]
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(x_values, y_line, marker='o', color='blue', linestyle='-')
axes[0].set_title('Left: Line Plot')
axes[0].set_xlabel('Category')
axes[0].set_ylabel('Line Value')
axes[0].grid(True, linestyle='--', alpha=0.6)
axes[1].bar(x_values, y_bar, color='green', alpha=0.7)
axes[1].set_title('Right: Bar Plot')
axes[1].set_xlabel('Category')
axes[1].set_ylabel('Bar Value')
plt.tight_layout()
plt.show()
```

```python
a = [12, 15, 14, 10, 18, 20, 19, 16, 17]
plt.boxplot(a)
plt.title("Boxplot graph:")
plt.ylabel("Values")
plt.show()
```

```python
a = ['A', 'B', 'C']
b = [5,8,4]
plt.barh(a,b,color='gray')
plt.show()
```

```python
group1 = [3, 5, 7]
group2 = [2, 4, 6]
categories = ['Category 1', 'Category 2', 'Category 3']
x_positions = np.arange(len(categories))
plt.figure(figsize=(8, 6))
plt.bar(x_positions, group1, label='Group 1', color='skyblue', edgecolor='black')
plt.bar(x_positions, group2, bottom=group1, label='Group 2', color='darkorange', edgecolor='black')
plt.xlabel('Categories', fontsize=12)
plt.ylabel('Values', fontsize=12)
plt.title('Stacked Bar Chart for Group 1 and Group 2', fontsize=14, fontweight='bold')
plt.xticks(x_positions, categories, fontsize=10)
plt.legend(loc='upper left')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```
