```python
import pandas as pd
data = {"Product":['pen','laptop','mobile','grocery'],
        "price":[10,44000,15000,2500],
        "day":['sun','mon','tues','thurs']}
a = pd.DataFrame(data)
print(a)
a.to_excel("sales.xlsx", index=False)  # Use to_excel() method, not pd.to_excel()
print("\nData saved to sales.xlsx")
b = pd.read_excel("sales.xlsx")
print("\nData read from Excel:")
print(b)
```

```python
df = pd.DataFrame({"A": [1, 2, 3], "B": [4, 5, 6], "C": [7, 8, 9]})
df.rename(columns={"A": "Maths", "B": "Science", "C": "English"}, inplace=True)
print(df)
```

```python
df = pd.DataFrame({"Name": ["A", "B", "C"], "Age": [22, 18, 25], "Address": ["X", "Y", "Z"]})
df.drop(columns=["Address"], inplace=True)
print(df)
```

```python
df = pd.DataFrame({
    "Name": ["A", "B", "C"],
    "Age": [22, 18, 25],
    "Marks": [80, 60, 90]
})
filtered_df = df[(df["Age"] > 20) & (df["Marks"] > 70)]
print(filtered_df)
```

```python
df = pd.DataFrame({"Name": ["A", "B", "C"], "Age": ["22", "18", "25"]})
print(f"Before: {df['Age'].dtype}")
df["Age"] = df["Age"].astype(int)
print(f"After: {df['Age'].dtype}")
print(df)
```

```python
df = pd.DataFrame({
    "A": [1, None, 3, None],
    "B": [None, 5, 6, 7],
    "C": [8, 9, None, 11]
})
missing_counts = df.isnull().sum()
print("Missing values per column:")
print(missing_counts)
missing_percentage = (df.isnull().sum() / len(df)) * 100
print("\nMissing percentage:")
print(missing_percentage)
```

```python
df = pd.DataFrame({
    "Age": [22, 18, 25, 30, 21],
    "Marks": [80, 60, 90, 85, 75],
    "Salary": [30000, 25000, 40000, 35000, 32000]
})
print("Summary statistics:")
print(df.describe())
print("\nMean values:")
print(df.mean())
print("\nStandard deviation:")
print(df.std())
```

```python
df = pd.DataFrame({
    "Quantity": [2, 5, 3, 4],
    "Price": [100, 50, 200, 150]
})
df["Total"] = df["Quantity"] * df["Price"]
print(df)
```

```python
df = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E"],
    "Marks": [80, 60, 90, 85, 75]
})
top_3 = df.nlargest(3, "Marks")
print("Top 3 marks:")
print(top_3)
top_3_values = df["Marks"].nlargest(3)
print("\nTop 3 marks values:")
print(top_3_values)
```

```python
df = pd.DataFrame({
    "City": ["Pune", "Pune", "Delhi", "Delhi", "Mumbai"],
    "Person": ["A", "B", "C", "D", "E"]
})
result = df.groupby("City")["Person"].count().reset_index()
result.columns = ["City", "Person_Count"] 
print(result)
result2 = df.groupby("City").size().reset_index(name="Count")
print("\nUsing size():")
print(result2)
```
