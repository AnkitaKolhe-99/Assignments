```python
import pandas as pd
import numpy as np
data = {"StudentID":[1,2,2,4,5,6,7,8,9,10],"Name":["Arav","Priya","Rohan","Sara","Vikram","Ananya","Kabir","Meera","Yash","Diya"],
       "Age":[16,17,16,np.nan,17,16,15,17,16,15],"Gender":['M','F','M','F','M','F','M','F','M','F'],"StudyHours":[2,3,1,4,5,2,3,4,1,2],
        "Attendance":[85,92,70,95,88,90,75,98,65,80],"ParentsGraduation":["Graduate","Postgraduate","Diploma","Graduate","Postgraduate","Graduate","Diploma",
        "Graduate","Diploma","Postgraduate"], "ExamScore":[72,88,60,95,88,90,75,98,65,80]}
df = pd.DataFrame(data)
df.to_csv("Student_analysis.csv",index=False)
print("Dataset created...")
```

```python
df = pd.read_csv("Student_analysis.csv")
df.head(6)
print(df.info())
print(df.describe())
print(df.shape)
```

```python
#missing data
print(df.isnull())
print(df.isnull().sum())
df['Age'] = df['Age'].fillna(df['Age'].mean())
df.to_csv("Students_analysis.csv",index=False)
```

```python
# remove duplicates
print("Duplicated:",df['StudentID'].duplicated().sum())
df['StudentID'] = df['StudentID'].drop_duplicates()
df.to_csv("Students_analysis.csv",index=False)
```

```python
print(df.dtypes)
df['Age'] = df['Age'].astype(int)
print(df.dtypes)
df.to_csv("Students_analysis.csv",index=False)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.histplot(df['Age'], kde=True)
plt.title('Age distribution')
plt.show()
```

```python
# Average score by parent education
sns.barplot(x='ParentsGraduation',y='ExamScore', data=df)
plt.title("Analysis")
plt.show()
```

```python
#attendance vs examscore
sns.scatterplot(x='Attendance',y='ExamScore',hue='Gender' ,data=df)
plt.title("Attendance vs ExamScore")
plt.show()
```

```python
# gender count
sns.countplot(x='Gender',data=df)
plt.title("Gender Count")
plt.show()
```

```python
# Bivarient analysis
import seaborn as sns
import matplotlib.pyplot as plt
sns.scatterplot(x='StudyHours',y='ExamScore',hue='Gender', data=df)
plt.title("study hours vs exam score")
plt.show()
```
