```python
df['tip_percent'] = df['tip'] / df['total_bill'] * 100
df['tip_percent']
```

```python
df["bill_category"] = np.select([df["total_bill"] < 15, (df["total_bill"] >= 15) & (df["total_bill"] <= 30), df["total_bill"] > 30], 
        ["Low", "Medium", "High"], default="Unknown")
df["bill_category"]
```

```python
df = df.sort_values(by="tip_percent", ascending=False)
df
```

```python
df = df.sort_values(by=["day", "total_bill"], ascending=[True, False])
df
```

```python
df.isnull().sum()
```

```python
mv = df['total_bill'].mean()
df['total_bill'] = df['total_bill'].fillna(mv)
df['total_bill']
```

```python
s = df['size'].median()
df['size'] = df['size'].fillna(s)
df['size']
```

```python
s = df['tip_percent'].mean()
df['tip_percent'] = df['tip_percent'].fillna(s)
df['tip_percent']
```

```python
df['sex'] = df['sex'].fillna(df['sex'].mode()[0])
```

```python
df = df.drop(columns=['smoker'])
df
```

```python
df.isnull().sum()
```

```python
df.to_csv('modified_tips',index=False)
```
