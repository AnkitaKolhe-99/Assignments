```python
import pandas as pd
df = pd.DataFrame({
    "Name": ["Amit", "Rahul", "Sneha", "Priya", "Karan"],
    "Marks": [85, 67, 92, 74, 88],
    "Age": [20, 19, 21, 18, 22]
})
# Marks greater than 90
print(df[df["Marks"] > 90])
# Age less than 20
print(df[df["Age"] < 26])
# Multiple conditions
print(df[(df["Marks"] > 85) & (df["Age"] < 32)])
```

```python
import pandas as pd

# Create DataFrame
df = pd.DataFrame({
    "Name": ["Amit", "Rahul", "Sneha", "Priya"],
    "Maths": [85, 78, 92, 88],
    "Science": [80, 75, 95, 90]
})

print(df)

df["Total"] = df["Maths"] + df["Science"]

print(df)

df["Percentage"] = df["Total"] / 2

print(df)

df["Grade"] = df["Percentage"].apply(
    lambda x: "A" if x >= 90 else "B"
)

print(df)
```
