### 1. What is the difference between a Pandas Series and DataFrame?

A **Series** is a one-dimensional labeled array. A **DataFrame** is a two-dimensional labeled table containing rows and columns. A DataFrame can contain multiple Series as its columns.

### 2. How do you efficiently load large CSV or Excel files into Pandas?

For large CSV files, use options such as `usecols`, `dtype`, and `chunksize` to reduce memory usage and process data in pieces. For Excel, select only required sheets/columns where possible. CSV is generally more efficient than Excel for very large datasets.

### 3. How can you handle missing or inconsistent data in a real dataset?

First identify missing values with `isna()`/`isnull()`. Then decide whether to remove rows/columns or fill values using methods such as median, mean, mode, forward fill or backward fill. Inconsistent values should be standardized, for example by trimming whitespace, fixing capitalization, and converting data types.

### 4. Explain how to filter rows and columns in a DataFrame.

Use boolean conditions for rows, for example `df[df['Revenue'] > 5000]`. Use column selection such as `df[['Product', 'Revenue']]`. For label-based selection use `loc`, and for integer-position selection use `iloc`.

### 5. When and why would you use groupby in a DataFrame?

Use `groupby()` when you need to split data into groups and calculate an aggregate or transformation for each group. For example, `df.groupby('Product')['Revenue'].mean()` calculates average revenue for each product.

### 6. What is the difference between merge, join, and concat operations in Pandas?

`merge()` combines DataFrames using one or more matching keys, similar to SQL joins. `join()` is mainly convenient for combining DataFrames using their indexes. `concat()` stacks or combines DataFrames along rows or columns.

### 7. How can you calculate correlation between two numeric columns?

Use `df['Revenue'].corr(df['Profit'])`. You can also use `df.corr()` to calculate correlations among numeric columns. The correlation coefficient generally ranges from -1 to +1.

### 8. Explain how pivot tables work in Pandas and give a use case.

A pivot table summarizes data by one or more dimensions using an aggregation function. For example, `pd.pivot_table(df, values='Revenue', index='Region', columns='Product', aggfunc='sum')` can show total revenue for every product in each region.

### 9. How can you sort a DataFrame by multiple columns?

Use `sort_values()`. For example: `df.sort_values(by=['Region', 'Revenue'], ascending=[True, False])` sorts Region in ascending order and Revenue in descending order.

### 10. How would you detect and remove duplicates in real-world data?

Detect duplicates with `df.duplicated()` or `df.duplicated(subset=['EmployeeID'])`. Remove them with `df.drop_duplicates()` or `df.drop_duplicates(subset=['EmployeeID'])` when a specific key defines uniqueness.

```python
import pandas as pd
data = {'Name':['John','Alice','Bob'],'Age':[25,None,30]}
df = pd.DataFrame(data)
print(df.fillna(df['Age'].median()))
```

```python
import pandas as pd
data = {'Revenue':[1000,1500,2000],'Profit':[200,300,400]}
df = pd.DataFrame(data)
print(df.corr())
```

**Output:** Revenue and Profit have a perfect positive correlation of **1.0**, because Profit is exactly 20% of Revenue.

```python
import pandas as pd
data = {'EmployeeID':[1,2,3],'Name':['John','Alice','Bob'],'Salary':[5000,6000,5500]}
df1 = pd.DataFrame(data)
df2 = pd.DataFrame({'EmployeeID':[1,2,3],'Bonus':[500,600,550]})
merged_df = pd.merge(df1,df2,on='EmployeeID')
print(merged_df)
```

**Output:** The merge creates one row per EmployeeID with columns EmployeeID, Name, Salary and Bonus.
