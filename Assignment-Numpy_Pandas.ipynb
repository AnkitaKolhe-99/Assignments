```python
import pandas as pd
data = {"Name":['Naman','Sagar','Shubham','Veer','Nirman'],"Age":[22,24,32,21,19],"Marks":[90,95,75,87,69]}
a = pd.DataFrame(data)
print(a)
```

```python
df = a.assign(grades=['A','B','C','D','F'])
print(df)
```

```python
a.loc[a['Marks'] >= 90, 'Grade'] = 'A'
a.loc[(a['Marks'] >= 70) & (a['Marks'] < 80), 'Grade'] = 'B'
a.loc[(a['Marks'] >= 60) & (a['Marks'] < 70), 'Grade'] = 'C'
print(a)
```
