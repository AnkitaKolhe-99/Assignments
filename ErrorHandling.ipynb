```python
class InvalidAgeError(Exception):
    def __init__(self,age,message = 'Age not valid. Please recheck'):
        self.age = age
        self.message = message
def chkAge_voting(age):
    if age < 18 or age > 131:
        raise InvalidAgeError(age)
    return "Eligible Voter"
def chk_schoolAdmission(age):
    if age < 4 or age > 7:
        raise InvalidAgeError(age)
    return "Eligible for pre-school admission"
print("Check age criteria for - ")
print("1. Voting Right")
print("2. Pre-school Admission")
ch = int(input("Enter your choice..."))
if ch == 1:
    try:
        a = int(input("Enter the Age: "))
        result1 = chkAge_voting(a)
        print(result1)
    except ValueError:
        print("Enter numbers only")
    except InvalidAgeError as e:
        print(f'Error: {e.message} you entered {e.age}')
elif ch == 2:
    try:
        b = int(input("Enter the Age: "))
        result2 = chk_schoolAdmission(b)
        print(result2)
    except ValueError:
        print("Enter the numbers only")
    except InvalidAgeError as f:
        print(f'Error: {f.message} you entered {f.age}')    
else:
    print("Invalid choice")
```
