```python
nu = 1221
rev = 0
c = nu
while nu > 0:
    d = nu % 10
    rev = rev * 10 + d
    nu //= 10
    
if c == rev:
        print("Palindrome")
else:
        print("Not Palindrome")    
    
print(rev)
```
